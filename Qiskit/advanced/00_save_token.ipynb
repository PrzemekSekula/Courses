# Initial configuration

Reads the token from keyboards and saves it to `$HOME/.qiskit/qiskit-ibm.json`.

To get the token go to quantum.ibm.com

```python
from IPython.display import display
import ipywidgets as widgets
from qiskit_ibm_runtime import QiskitRuntimeService
```

```python

password_input = widgets.Password(
    placeholder='Enter API Token',
    description='API Token:',
    disabled=False
)

button = widgets.Button(description="Submit")
output = widgets.Output()

# Event handler
def handle_submit(b):
    global token
    token = password_input.value 

    QiskitRuntimeService.save_account(
    token=token,
    overwrite=True,
    channel="ibm_quantum" # `channel` distinguishes between different account types
    )    

    with output:
        output.clear_output()
        # Use the token securely without printing it
        print(f"Token saved in $HOME/.qiskit/qiskit-ibm.json")

button.on_click(handle_submit)

# Display widgets
display(password_input, button, output)
```
