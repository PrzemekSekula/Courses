# Shor's algorithm basics

An implementation of a classical algorithm for factoring large numbers.



```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 11 * 13 # Number to factor
print(f"N = {N}")
a = 2 # Random number

my_gcd = math.gcd(a, N) # Check if a and N are coprime
if my_gcd != 1:
    print(f"a = {a} and N = {N} are not coprime")
    print(f"The Greatest Common Divisor is {my_gcd}")


print(f"a = {a} and N = {N} are coprime")





```

```python
PLOT_A_MOD_N = True

z_list = list(range(1, N))
y_list = [a**z % N for z in z_list]

if PLOT_A_MOD_N:
    plt.plot(z_list, y_list)
    plt.xlabel(f'{a}^z mod {N}') 
    plt.show()



print (z_list)
print (y_list)







```

```python
r = y_list.index(1)+1
print (f'r = {r}')

if r % 2 == 0:
    x = int((a**(r/2)) % N)
    print (f'x = {x}')
    if (x+1) % N != 0:
        print (f'x+1={x+1}, x-1={x-1}')
        gcd_x1 = math.gcd(x+1, N)
        gcd_x2 = math.gcd(x-1, N)
        print (f'gcd({x+1}, {N}) = {gcd_x1}, {N}/{gcd_x1} = {N/gcd_x1}')
        print (f'gcd({x-1}, {N}) = {gcd_x2}, {N}/{gcd_x2} = {N/gcd_x2}')
    else:
        print (f'x + 1 is 0 mod {N}')






```
